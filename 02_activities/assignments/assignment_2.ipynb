#### Starter Code for Question 1

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def is_duplicate(root: TreeNode) -> int:
    if not root:
        return -1

    seen = set()
    queue = [root] 

    while queue:
        node = queue.pop(0)  
        if node.val in seen:
            return node.val
        seen.add(node.val)
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
        return -1
```

```python
# Your answer here
def is_valid_brackets(s: str) -> bool:
    k_v_pairs = {
        '{':'}',
        '[':']',
        '(':')'
    }
    stack = []

    for c in s:
        #check if the character is in the valid key dictionary and add it to the stack 
        if c in k_v_pairs:
            stack.append(c)
        #check if the character is in the valid values dictionary and find key
        elif c in k_v_pairs.values():
            #check if the stack has any characters and if the last character in the stack corresponds to the key:value pair of the last opened bracket
            if len(stack) == 0 or c != k_v_pairs[stack[-1]]:
                return False
            stack.pop()
        #Every other character will return False
        else: return False
    return True
```
